## Example code: finding the most common word in a poem

My guess before running the code: the most common word will be "a" or "more".

```python
from collections import Counter

poem = "Shall I compare thee to a summer's day? Thou art more lovely and more temperate: Rough winds do shake the darling buds of May, And summer's lease hath all too short a date:"
words = [w.lower() for w in poem.split()]
counts = Counter(words)
most_common = counts.most_common(1)[0]
print("Most frequent word:", most_common[0], "appears", most_common[1], "times")
```

**Question to Copilot:** Why can't a browser play .mid files but can play .ogg files?

**Answer:** An .ogg file contains encoded audio that the browser can decode and play directly. A .mid (MIDI) file contains musical instructions (which notes to play, which instrument, how long and how loud), more like sheet music than a recording. **Question to Copilot:** Why does multiplying a picture's pixel array by 2/3 change how it looks?

## sklearn concept 1: Loading datasets

**Prompt:** I am an absolute beginner in coding. Please give me an example demonstrating how to load a built-in dataset using the sklearn datasets library, with a step-by-step explanation.

**Ambiguous terms I noted:** (skriv 1-2 ord du inte förstod, t.ex. "Bunch object")

```python
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()

# Display the available parts of the dataset
print("Dataset contents:")
print(iris.keys())

# Display the feature names
print("\nFeature names:")
print(iris.feature_names)

# Display the species names
print("\nSpecies names:")
print(iris.target_names)

# Display the first five rows of data
print("\nFirst five flowers:")
print(iris.data[:5])

# Display the first five labels
print("\nFirst five labels:")
print(iris.target[:5])

# Display information about the first flower
print("\nFirst flower measurements:")
print(iris.data[0])

# Display the species name of the first flower
print("\nFirst flower species:")
print(iris.target_names[iris.target[0]])

# Display a description of the dataset
print("\nDataset description:")
print(iris.DESCR)
```

**Learning outcome:** Built-in datasets are loaded with functions that start with `load_`. The result is a dictionary-like object holding the data, labels and descriptions.

## sklearn concept 3: Exploring and visualizing data

**Prompt:** I am an absolute beginner in coding. Please show me how to calculate basic statistics and make a simple plot of the iris dataset from sklearn, with a step-by-step explanation.

```python
import numpy as np
from matplotlib import pyplot

print("Average of each feature:", np.mean(iris.data, axis=0))
print("Smallest values:", np.min(iris.data, axis=0))
print("Largest values:", np.max(iris.data, axis=0))

pyplot.scatter(iris.data[:, 0], iris.data[:, 1], c=iris.target)
pyplot.xlabel(iris.feature_names[0])
pyplot.ylabel(iris.feature_names[1])
pyplot.show()
```

## Pandas concept 1: Importing and data structures (Series and DataFrame)

**Prompt:** I am an absolute beginner in coding. Please explain how to import pandas and what a Series and a DataFrame are, with simple examples and a step-by-step explanation.

```python
# Import the pandas library and give it the nickname "pd"
import pandas as pd

# -------------------------------
# PART 1: Creating a Series
# -------------------------------

print("SERIES EXAMPLE")
print()

# A Series is like a single column of data
numbers = pd.Series([10, 20, 30, 40])

print("The Series:")
print(numbers)

print()
print("First value in the Series:")
print(numbers[0])

print("\n" + "-" * 40 + "\n")

# -------------------------------
# PART 2: Creating a DataFrame
# -------------------------------

print("DATAFRAME EXAMPLE")
print()

# A DataFrame is like a spreadsheet table
people = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 22]
})

print("The DataFrame:")
print(people)

print("\n" + "-" * 40 + "\n")

# -------------------------------
# PART 3: Accessing a column
# -------------------------------

print("NAME COLUMN")
print()

# Selecting one column returns a Series
names = people["Name"]

print(names)

print("\n" + "-" * 40 + "\n")

# -------------------------------
# PART 4: Adding a new column
# -------------------------------

print("ADDING A NEW COLUMN")
print()

people["Country"] = ["UK", "Canada", "Sweden"]

print(people)

print("\n" + "-" * 40 + "\n")

# -------------------------------
# PART 5: Pet Data Example
# -------------------------------

print("PET DATA")
print()

pets = pd.DataFrame({
    "Name": ["Max", "Bella", "Charlie"],
    "Species": ["Dog", "Cat", "Dog"],
    "Age": [3, 5, 2]
})

print(pets)

print()
print("Pet names only:")
print(pets["Name"])
```

**Learning outcome:** `import pandas as pd` loads the library with the short name `pd`. A Series is a single column of labelled data, and a DataFrame is a whole table made of several Series (columns).

## Pandas concept 2: Creating DataFrames

**Prompt:** I am an absolute beginner in coding. Please show me how to create a pandas DataFrame from a dictionary, from a list, from a CSV file and from a scikit-learn dataset, with a step-by-step explanation.

```python
# ==========================================
# PANDAS DATAFRAME EXAMPLES
# ==========================================

# Import pandas
import pandas as pd

# ==========================================
# 1. DATAFRAME FROM A DICTIONARY
# ==========================================

print("1. DATAFRAME FROM A DICTIONARY")
print()

dictionary_data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 22]
}

df_dictionary = pd.DataFrame(dictionary_data)

print(df_dictionary)

print("\n" + "=" * 50 + "\n")

# ==========================================
# 2. DATAFRAME FROM A LIST
# ==========================================

print("2. DATAFRAME FROM A LIST")
print()

list_data = [
    ["Alice", 25],
    ["Bob", 30],
    ["Charlie", 22]
]

df_list = pd.DataFrame(
    list_data,
    columns=["Name", "Age"]
)

print(df_list)

print("\n" + "=" * 50 + "\n")

# ==========================================
# 3. DATAFRAME FROM A CSV FILE
# ==========================================

print("3. DATAFRAME FROM A CSV FILE")
print()

# Create a sample CSV file
csv_text = """Name,Age
Alice,25
Bob,30
Charlie,22
"""

with open("people.csv", "w") as file:
    file.write(csv_text)

# Read the CSV file into a DataFrame
df_csv = pd.read_csv("people.csv")

print(df_csv)

print("\n" + "=" * 50 + "\n")

# ==========================================
# 4. DATAFRAME FROM A SCIKIT-LEARN DATASET
# ==========================================

print("4. DATAFRAME FROM THE IRIS DATASET")
print()

from sklearn.datasets import load_iris

# Load the dataset
iris = load_iris()

# Create a DataFrame from the feature data
df_iris = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add the target values as a new column
df_iris["species"] = iris.target

# Display the first five rows
print(df_iris.head())

print("\n" + "=" * 50 + "\n")

# ==========================================
# EXTRA INFORMATION
# ==========================================

print("DATAFRAME SHAPES")
print()

print("Dictionary DataFrame:", df_dictionary.shape)
print("List DataFrame:", df_list.shape)
print("CSV DataFrame:", df_csv.shape)
print("Iris DataFrame:", df_iris.shape)
```
